```python
import pandas as pd
```

```python
URL = 'https://raw.githubusercontent.com/hovhannisyan91/sql-analytics-portfolio/refs/heads/main/data/public_schema/'

sql_path = 'C:/aca/sql-analytics-portfolio/Data/public_schema/'

l = ["customers", "employees", "orders", "products", "sales"]

for i in l:
    
    print(100*'-')
    
    df_name = URL + i + '.csv'
    print(f"trying to download {i}.csv")

    df = pd.read_csv(df_name)
    print(f"{i}.csv is downloaded") 
    
    print(f"the shape of {i}.csv : {df.shape}")
    
    print(f"trying to save  {i}.csv")
    df.to_csv(sql_path + i + '.csv', index=False)
    print(f"{i}.csv is successfuly saved")
    
```
