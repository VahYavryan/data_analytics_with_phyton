## Part 1

### df_products + df_departrments

```python
df_departments = pd.read_csv('../data/processed/departments_t.csv')
print(df_departments.shape)
df_departments.head()
```

```python
df_products = pd.read_csv('../data/raw/products.csv')
print(df_products.shape)
df_products.head()
```

```python
# Merge

df_departments_products = pd.merge(
    df_departments,
    df_products,
    on='department_id',
    how='left'
)
```

```python
df_departments_products.head()
```

### Քանի՞ ապրանք կա յուրաքանչյուր բաժնում (department)

```python

df_departments_products["department"].value_counts()

```

```python
df_departments_products["product_name"].describe()

```

```python
df_departments_products.info()
```

### df_products_departments + df_aisles

```python
df_aisles = pd.read_csv('../data/raw/aisles.csv')
print(df_aisles.shape)
df_aisles.head()
```

```python
# Merge

df_products_enriched = pd.merge(
    df_departments_products,
    df_aisles,
    on='aisle_id',
    how='left'
)

df_products_enriched.head()
```

```python
# Cheking _merge
df_products_enriched_check = pd.merge(
    df_departments_products,
    df_aisles,
    on='aisle_id',
    how='left',
    indicator=True
)

df_products_enriched_check['_merge'].value_counts()
```
