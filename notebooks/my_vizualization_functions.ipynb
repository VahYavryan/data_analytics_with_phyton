### Import Packages

```python
import plotly.express as px
import pandas as pd
```

### My Bar Chart Function

```python
def my_bar_chart(data, x, y, top=True):
    sorted_data = data.sort_values(by=y, ascending=not top)
    
    fig = px.bar(
        sorted_data, 
        x=x, 
        y=y, 
        barmode="group",
        color_discrete_sequence=["#245275"],
        title=f"Bar Chart: {y.capitalize()} over {x.capitalize()}",
        labels={
            x: x.capitalize(),
            y: y.capitalize()   
        }
    )
    
    fig.update_layout(
        plot_bgcolor="#F8F9FA",   
        paper_bgcolor="#FFFFFF",  
        
        xaxis=dict(showgrid=False),
        yaxis=dict(showgrid=True, gridcolor="#E5E7EB"))
    
    return fig 
```

### My List Chart Function

```python
def my_line_chart(data, x, y):
    fig = px.line(
        data, 
        x=x, 
        y=y, 
        markers=True,
        color_discrete_sequence=["#830000"],
        title=f"Line Chart: {y.capitalize()} over {x.capitalize()}",
        labels={
            x: x.capitalize(),
            y: y.capitalize()   
        })
    
    fig.update_layout(
        plot_bgcolor="#F8F9FA",   
        paper_bgcolor="#FFFFFF",  
        
        xaxis=dict(showgrid=False),
        yaxis=dict(showgrid=True, gridcolor="#E5E7EB"))
    
    return fig 
```
