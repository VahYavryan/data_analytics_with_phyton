### Data Wrangling և Subsetting

#### Սյուների անվանափոխում

```python
df_orders_new = df_orders.rename(columns={"order_dow": "order_days_of_week"})
```

```python
df_orders.rename(columns={"order_dow": "order_days_of_week"}, inplace=True)
```

#### Տվյալների տիպերի փոփոխում

```python
df_orders_new = df_orders.astype({"order_id": "int64"})
```

```python
df_orders.astype({"order_id": "int64"})
```

```python
df_orders["order_id"] = df_orders["order_id"].astype("int64")
```

#### Տվյալների փոխակերպում

```python
df_orders["is_weekend"] = df_orders["order_days_of_week"].isin([0, 6])
df_orders["is_morning"] = df_orders["order_hour_of_day"] < 12
```

```python
df_orders["order_frequency_category"] = df_orders["order_number"].apply(
    lambda x: "New" if x == 1
    else "Low" if x <= 5
    else "High"
)
```

```python
df_orders["days_since_prior_order"] = df_orders["days_since_prior_order"].fillna(0)
```

#### DataFrame Transpose (շրջում)

```python
# load original
df_departments_raw = pd.read_csv('../data/raw/departments.csv')

# transpose
df_t = df_departments_raw.T

# reset index
df_t = df_t.reset_index()

# rename columns
df_t.columns = ['department_id', 'department']

# remove first row (header-like row)
df_t = df_t[df_t['department_id'] != 'department_id']

# convert types
df_t['department_id'] = df_t['department_id'].astype(int)

# reset index
df_t = df_t.reset_index(drop=True)

df_t.head()
```

```python
df_t.to_csv('../data/processed/departments_t.csv', index=False)
```
