# Data Vizualization | Plotly

### Import Pakcages

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

## Dummy Dataset for Plotly Examples

```python
np.random.seed(42)
px.defaults.template = "plotly_white"

months = pd.date_range("2024-01-01", periods=6, freq="MS")
regions = ["North", "South", "East"]
products = ["A", "B"]

rows = []

for month in months:
    for region in regions:
        for product in products:
            sales = np.random.randint(80, 220)
            customers = np.random.randint(20, 90)
            units = np.random.randint(10, 70)
            
            rows.append([month, region, product, sales, customers, units])
            
df_plotly = pd.DataFrame(
    rows,
    columns=["month", "region", "product", "sales", "customers", "units"]
)

df_plotly["month_name"] = df_plotly["month"].dt.strftime("%b")
df_plotly.head()
```

### Bar Plot

```python
region_sales = (
    df_plotly.groupby("region", as_index=False)["sales"]
             .sum())
region_sales
```

```python
fig = px.bar(
    region_sales,
    x="region",
    y="sales",
    text="sales",
    title="Total Sales by Region"
)

fig.update_traces(textposition="outside")
fig.update_layout(
    xaxis_title="Region",
    yaxis_title="Sales"
)

fig.show()
```

### Line Plot

```python
monthly_sales = (
    df_plotly.groupby("month", as_index=False)["sales"]
    .sum())

monthly_sales
```

```python
fig = px.line(
    monthly_sales,
    x="month",
    y="sales",
    markers=True,
    title="Monthly Total Sales")

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales")

fig.show()
```

### Histogram

```python
fig = px.histogram(
    df_plotly,
    x="sales",
    nbins=15,
    title="Distribution of Sales")
fig.update_layout(
        xaxis_title="Sales",
        yaxis_title="Count")
fig.show()
```

### Scatter Plot

```python
fig = px.scatter(
    df_plotly,
    x="customers",
    y="sales",
    color="region",
    title="Customers VS Sales",
    hover_data=["product", "month_name", "units"]
)

fig.update_layout(
    xaxis_title="Customers",
    yaxis_title="Sales",
    legend_title="Region"
)

fig.show()
```

### Multi-Line Chart

```python
monthly_region_sales = (
    df_plotly.groupby(["month", "region"], as_index=False)["sales"]
    .sum())
monthly_region_sales.head()
```

```python
fig = px.line(
    monthly_region_sales,
    x="month",
    y="sales",
    color="region",
    markers=True,
    title="Monthly Sales by Region"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales",
    legend_title="Region"
)

fig.show()
```

### Grouped Bar Plot

```python
region_product_sales = (
    df_plotly.groupby(["region", "product"], as_index=False)["sales"]
    .sum()
)

region_product_sales
```

```python
fig = px.bar(
    region_product_sales,
    x="region",
    y="sales",
    color="product",
    barmode="group",
    title="Sales by Region and Product")

fig.update_layout(
    xaxis_title="Region",
    yaxis_title="Sales",
    legend_title="Product"
)

fig.show()
```

### Donut Chart

```python
product_sales = (
    df_plotly.groupby("product", as_index=False)["sales"]
    .sum())
product_sales
```

```python
fig = px.pie(
    product_sales,
    names="product",
    values="sales",
    hole=0.5,
    title="Share of Total Sales by Product")
fig.update_traces(textinfo="label+percent")
fig.show()
```

## Vertical and Horizontal Reference Lines


### Horizontal Lines

```python
monthly_sales = (
    df_plotly.groupby("month", as_index=False)["sales"]
    .sum())
monthly_sales
```

```python
avg_sales = monthly_sales["sales"].mean()
avg_sales
```

```python
fig = px.line(
    monthly_sales,
    x="month",
    y="sales",
    markers=True,
    title="Line Plot with Horizontal Reference Line")
fig.add_hline(
    y=avg_sales,
    line_dash="dot",
    annotation_text="Average Sales",
    annotation_position="top left")
fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales")
fig.show()
```

### Vertical Lines

```python
campaign_start = pd.Timestamp("2024-04-01")

fig = px.line(
    monthly_sales,
    x="month",
    y="sales",
    markers=True,
    title="Line Plot with Vertical Event Line"
)

fig.add_vline(
    x=campaign_start,
    line_dash="dot",
    line_color="red"
)

fig.add_annotation(
    x=campaign_start,
    y=monthly_sales["sales"].max(),
    text="Campaign Start",
    showarrow=True,
    arrowhead=2,
    ax=40,
    ay=-40
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales"
)

fig.show()
```

### Horizontal and Vertical Lines Together


```python
campaign_start = pd.Timestamp("2024-04-01")
avg_sales = monthly_sales["sales"].mean()

fig = px.line(
    monthly_sales,
    x="month",
    y="sales",
    markers=True,
    title="Line Plot with Horizontal and Vertical Reference Lines"
)

fig.add_hline(
    y=avg_sales,
    line_dash="dash",
    annotation_text="Average Sales",
    annotation_position="top left"
)

fig.add_vline(
    x=campaign_start,
    line_dash="dot",
    line_color="red"
)

fig.add_annotation(
    x=campaign_start,
    y=monthly_sales["sales"].max(),
    text="Campaign Start",
    showarrow=True,
    arrowhead=2,
    ax=40,
    ay=-40
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales"
)

fig.show()
```
