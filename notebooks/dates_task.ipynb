## Date Arithmetic

### Calculate the difference between `end` and `start`

```python
start = date(2025, 1, 10)
end = date(2025,2,5)
```

```python
delta = end - start
type(delta)
```

```python
print(type((delta).days))
delta
```

```python
print(start + timedelta(days=10))
print(start - timedelta(days=14))
print(start + timedelta(weeks=3))
```

```python
niceday = datetime(2026, 5, 22, 12, 1, 1)
```

```python
print(niceday + timedelta(hours=5, minutes=20, seconds=45))
```

## Practice with Pandas Dates

```python
df_dataframe = pd.DataFrame({
    "customer_id": [1, 2, 3, 4, 5],
    "signup_date": ["2024-1-10", "2024-2-5", "2024-3-12", "2024-4-1", "2024-5-20"],
    "last_purchase": ["2024-2-15", "2024-3-01", "2024-3-28", "2024-5-10", "2024-6-25"]
})

df_dataframe.info()
```

```python
df_dataframe.to_csv("df_dataframe.csv", index=False)
```

```python
pd.read_csv("df_dataframe.csv")
```

```python
df_dataframe["signup_date"] = pd.to_datetime(df_dataframe["signup_date"])
df_dataframe["last_purchase"] = pd.to_datetime(df_dataframe["last_purchase"])
```

```python
df_dataframe.info()
```

### Extract Date Components

```python
df_dataframe["signup_year"] = df_dataframe["signup_date"].dt.year
df_dataframe["signup_quarter"] = df_dataframe["signup_date"].dt.quarter
df_dataframe["signup_month"] = df_dataframe["signup_date"].dt.month
df_dataframe["signup_day"] = df_dataframe["signup_date"].dt.day
df_dataframe["signup_month_name"] = df_dataframe["signup_date"].dt.month_name()
df_dataframe["signup_day_name"] = df_dataframe["signup_date"].dt.day_name()
```

```python
df_dataframe
```

```python
df_dataframe["signup_dow"] = df_dataframe["signup_date"].dt.day_of_week
```

```python
df_dataframe['is_weekend'] = df_dataframe['signup_dow'].isin([0, 6]).astype(int)
```

```python
df_dataframe
```

```python
df_dataframe["days_between_signup_and_purchase"] = (df_dataframe["last_purchase"] - df_dataframe["signup_date"]).dt.days
```

```python
df_dataframe.head()
```

### Filter Data

```python
filtered_df = df_dataframe[df_dataframe["signup_date"] > "2024-03-01"].sort_values(by= "signup_date", ascending=True)
filtered_df
```

```python
filtered_by_march_df = df_dataframe[df_dataframe["signup_month"] == 3].sort_values(by= "signup_date", ascending=True)
filtered_by_march_df
```

```python
filtered_by_year_df = df_dataframe[df_dataframe["signup_year"] == 2024].sort_values(by= "signup_date", ascending=True)
filtered_by_year_df
```

```python
df_dataframe_1["date"] = pd.to_datetime(df_dataframe_1["date"])
```

```python
df_dataframe_1['year'] = df_dataframe_1['date'].dt.year
df_dataframe_1['month'] = df_dataframe_1['date'].dt.month
df_dataframe_1['day'] = df_dataframe_1['date'].dt.day
df_dataframe_1['day_name'] = df_dataframe_1['date'].dt.day_name()
df_dataframe_1['dow'] = df_dataframe_1['date'].dt.dayofweek
df_dataframe_1['is_weekend'] = df_dataframe_1['date'].dt.dayofweek.isin([5, 6]).astype(int)
```

```python
df_dataframe_1.head()
```

### Beginning and End of Month

```python
df_dataframe_1['month_start'] = df_dataframe_1["date"] - MonthBegin(1)
```

```python
df_dataframe_1["month_end"] = df_dataframe_1["date"] - MonthEnd(0)
```
