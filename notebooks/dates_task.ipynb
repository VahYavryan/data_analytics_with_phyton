# DateTime

## Import Packages

```python
import pandas as pd
from datetime import datetime, date, timedelta
```

## Practice with the datetime Module

### Create a date Object

```python
my_date = date(2025, 3, 15)
my_date
```

```python
print (f"Year: {my_date.year}")
print(f"Month: {my_date.month}")
print(f"Day: {my_date.day}")
print(f"Date: {my_date}")
```

### Create a datetime Object

```python
meeting_time = datetime(2025, 8, 8, 16, 45, 30)
meeting_time
```

```python
print(f"Year: {meeting_time.year}")
print(f"Month: {meeting_time.month}")
print(f"Day: {meeting_time.day}")
print(f"Hour:-> {meeting_time.hour}")
print(f"Minute:-> {meeting_time.minute}")
print(f"Second:-> {meeting_time.second}")
```

### Format Dates as Strings using meeting_time

```python
meeting_time.strftime("%Y/%m/%d")
```

```python
meeting_time.strftime("%d/%m/%Y")
```

```python
meeting_time.strftime("%B %d, %Y")
```

### Parse a String into a Datetime

```python
dt = datetime.strptime("2025-12-31", "%Y-%m-%d")
dt
```

```python
type(dt)
```

```python
print(f"Date: {dt}")
```
