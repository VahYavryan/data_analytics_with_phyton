### Part 2

#### Loops

```python
# Sum
total = 0
for i in revenues:
    total += i

print(total)

# average
avg = total / len(revenues)

print(avg)

# max
max = revenues[0]
for i in revenues:
    if i > max:
        max = i

print(max)

# count > 150
high_revenues = 0
for i in revenues:
    if i > 150:
        high_revenues += 1

print(high_revenues)

# commission
commission = [i * 1.05 for i in revenues]
commission
```

### Part 3

#### if / elif / else

```python
revenues = [130, 170, 210, 75, 300]
```

```python
segments = []

for i in revenues:
    i = int(i)

    if i > 200:
        segments.append("Premium")

    elif 100 < i <= 200:
        segments.append("Standart")
    else: 
        segments.append("Low")

print(segments)
```
